```python
#Arithmetic Operators, Augmented Assignment Operator

#Restaurant

##
numOfPizzas = 2

numOfPizzas = numOfPizzas + 5
print(numOfPizzas)

numOfPizzas += 3
print(numOfPizzas)

##
numberofpersons = 10

numberofpersons = numberofpersons - 2
print(numberofpersons)

numberofpersons -= 3
print(numberofpersons)

##
priceOfPizza = 10

priceOfPizza = priceOfPizza * 3
print(priceOfPizza)

priceOfPizza *= 5
print(priceOfPizza)

##

shareTheBill = 500

shareTheBill = shareTheBill / 5
print(shareTheBill)

shareTheBill /=2
print(shareTheBill)
```
