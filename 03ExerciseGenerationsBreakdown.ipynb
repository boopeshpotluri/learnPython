```python
from datetime import datetime

#IfElse IfElIF IfElIfElse

yearOfBirth = input("Please enter your year of birth : ")

yearOfBirth = int(yearOfBirth)
currentYear = int(datetime.now().year)


if yearOfBirth>=2013 and yearOfBirth<=2026:
  print(f"Your current age is : {currentYear-yearOfBirth}. You belong to Generation Alpha !")
elif yearOfBirth>=1997 and yearOfBirth<=2012:
  print(f"Your current age is : {currentYear-yearOfBirth}. You belong to Generation Z !")
elif yearOfBirth>=1981 and yearOfBirth<=1996:
  print(f"Your current age is : {currentYear-yearOfBirth}. You belong to Millennials !")
elif yearOfBirth>=1965 and yearOfBirth<=1980:
  print(f"Your current age is : {currentYear-yearOfBirth}. You belong to Generation X !")
elif yearOfBirth>=1928 and yearOfBirth<=1964:
  print(f"Your current age is : {currentYear-yearOfBirth}. You belong to Baby Boomers & Silent Generation !")
else:
  print(f"Your current age is : {currentYear-yearOfBirth}. OMG you are AMAZING!! You would have walked with the Dinosaurs !")
```
