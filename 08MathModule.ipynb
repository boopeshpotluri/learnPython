```python
#Math Module

import math

print("The value of pi : ", math.pi)

print("Square root of 625 : ", math.sqrt(625));

x = 8.1
print("Ceil of 8.1 is :", math.ceil(x))

y = 8.8
print("Fllor of 8.8 is :", math.floor(y))

#Calculate Circumference  of a circle

radius = 10

circumference = 2 * math.pi * radius
print("Circumference of the circle with radius 10 is :", circumference)

```
