```python
#Logical Operators or, and, not

country = "America"

if(country == "Canada" or country == "America" or country == "Mexico"):
  print("Your choosen country is Canada/America/Mexico")

sport = "Ice Hockey"

if((country == "Canada" or country == "America") and sport == "Ice Hockey"):
  print("You choose country is Canada/America and sport is Ice Hockey")

residentCountry = "Japan"
if(not (residentCountry == "Canada" or residentCountry == "America" or residentCountry == "Mexico")):
  print("Your don't reside in Canada/America/Mexico")
```
