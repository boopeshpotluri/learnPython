```python
#Basic Calculator

operationType = input("Enter the type of arthematic operation you want to per + - * / :")

x = float(input("Enter the first number :"))
y = float(input("Enter the second number :"))

if(operationType == "+"):
  print(f"Adding {x} + {y} : {x+y}")
elif(operationType == "-"):
  print(f"Substracting {x} - {y} : {x-y}")
elif(operationType == "*"):
   print(f"Multiplying {x} * {y} : {x*y}")
elif(operationType == "/"):
  print(f"Dividing {x} / {y} : {x/y}")
else:
  print("You chose an incorrect operation...")
```
